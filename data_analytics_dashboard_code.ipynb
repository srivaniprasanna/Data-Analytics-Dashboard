```python
import streamlit as st
import pandas as pd
import time

# --- Page Config ---
st.set_page_config(
    page_title="Internship Dashboard",
    page_icon="🎯",
    layout="wide",
    initial_sidebar_state="expanded"
)

st.title("Internship Dashboard 🎯")

# --- File Upload Section ---
uploaded_file = st.file_uploader("Upload your Excel file (.xlsx)", type=["xlsx"])

if uploaded_file:
    # --- Loading Spinner + Progress Bar ---
    with st.spinner('Loading data... Please wait ⏳'):
        my_bar = st.progress(0)
        for percent_complete in range(0, 100, 10):
            time.sleep(0.05)
            my_bar.progress(percent_complete + 10)

        # Load Excel
        df = pd.read_excel(uploaded_file)

        # Clean column names
        df.columns = df.columns.str.strip().str.replace(' ', '_').str.lower()

    st.success('Data loaded successfully! ✅')

    # --- Show available columns
    st.subheader("🔎 Available Columns in Data:")
    st.write(df.columns.tolist())

    # --- Try to find relevant columns
    company_col = None
    role_col = None
    ctc_col = None

    for col in df.columns:
        if 'company' in col:
            company_col = col
        if 'role' in col:
            role_col = col
        if 'ctc' in col or 'stipend' in col:
            ctc_col = col

    if company_col and role_col:
        # Sidebar Filters
        with st.sidebar:
            st.title('Dashboard Controls ⚙️')
            st.subheader('Filters')

            unique_companies = df[company_col].dropna().unique()
            selected_company = st.selectbox('Select Company', options=['All'] + list(unique_companies))

            unique_roles = df[role_col].dropna().unique()
            selected_role = st.multiselect('Select Roles', options=unique_roles)

            min_ctc = int(df[ctc_col].min()) if ctc_col else 0
            max_ctc = int(df[ctc_col].max()) if ctc_col else 100000
            selected_ctc = st.slider('Select CTC Range', min_value=min_ctc, max_value=max_ctc, value=(min_ctc, max_ctc))

            st.write("---")
            apply_filters = st.button('Apply Filters')

        # Tabs
        tab1, tab2 = st.tabs(['📋 Internship Details', '📊 Summary Statistics'])

        with tab1:
            st.header('Internship Opportunities 📋')

            filtered_df = df.copy()

            if selected_company != 'All':
                filtered_df = filtered_df[filtered_df[company_col] == selected_company]

            if selected_role:
                filtered_df = filtered_df[filtered_df[role_col].isin(selected_role)]

            if ctc_col:
                filtered_df = filtered_df[
                    (filtered_df[ctc_col] >= selected_ctc[0]) &
                    (filtered_df[ctc_col] <= selected_ctc[1])
                ]

            st.dataframe(filtered_df, use_container_width=True)

        with tab2:
            st.header('Summary Statistics 📊')

            st.subheader('Number of Internships by Company')
            company_counts = df[company_col].value_counts()
            st.bar_chart(company_counts)

            if ctc_col:
                st.subheader('Average CTC Offered')
                avg_ctc = df[ctc_col].mean()
                st.metric(label="Average CTC", value=f"{avg_ctc:.2f}")

            st.subheader('Roles Available')
            role_counts = df[role_col].value_counts()
            st.write(role_counts)

        st.caption('Internship Dashboard | Powered by Streamlit 🚀')

    else:
        st.error("❌ Required columns ('company' and 'role') not found in the file! Please check your Excel sheet.")

else:
    st.info("👆 Please upload an Excel (.xlsx) file to begin.")
```
